```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import time
from sklearn.metrics import accuracy_score, confusion_matrix,mean_squared_error,mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import *
```

# A regression problem

```python
dataset = fetch_california_housing()
x = dataset.data
y = dataset.target
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=5361)
print(f'{X_train.shape =}')
print(f'{y_train.shape =}')
print(f'{X_test.shape =}')
print(f'{y_test.shape =}')
```

## Baseline model - linear regression

```python
model = LinearRegression()
model.fit(X_train,y_train)
pred = model.predict(X_test)
print('Mean squared error = {:.5f}'.format(mean_squared_error(pred,y_test)))
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.utils import to_categorical

# normalize inputs to 0-1 using training set min and max
x_min = np.min(X_train, axis=0)
x_max = np.max(X_train, axis=0)
x_range = x_max - x_min
x_range[x_range == 0] = 1

X_train_n = (X_train - x_min) / x_range
X_test_n = (X_test - x_min) / x_range

def neural_network_regression(n_inputs):
    model = tf.keras.models.Sequential()
    model.add(Input((n_inputs,)))
    model.add(Dense(64, activation='relu'))
    model.add(Dense(32, activation='relu'))
    model.add(Dense(1))
    return model

model = neural_network_regression(X_train_n.shape[1])
model.summary()
model.compile(optimizer=tf.keras.optimizers.Adam(), loss='mse', metrics=['mae'])

start_train = time.time()
history = model.fit(
    X_train_n, y_train,
    epochs=100,
    batch_size=500,
    verbose=2,
    validation_data=(X_test_n, y_test)
)
end_train = time.time()

pred = model.predict(X_test_n).reshape(-1)
print('Elapsed time training = {:.4f} secs'.format(end_train-start_train))
print('Mean squared error = {:.5f}'.format(mean_squared_error(pred, y_test)))
print('Mean absolute error = {:.5f}'.format(mean_absolute_error(pred, y_test)))
```

# A classification problem

```python
dataset = fetch_covtype()
x = dataset.data
y = dataset.target
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=5361)
print(f'{X_train.shape =}')
print(f'{y_train.shape =}')
print(f'{X_test.shape =}')
print(f'{y_test.shape =}')
print(f'{len(set(y_train)) = }')
```

## Baseline model - decision tree

```python
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
pred = model.predict(X_test)
print(f'{accuracy_score(pred,y_test) = :.4f}')
leaves = model.get_n_leaves()
print(f'The tree has {leaves-1} internal nodes and {leaves} leaves')
print(f"The tree's depth is {model.get_depth()}")
```

```python
# First neural network attempt

import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.utils import to_categorical

# normalize inputs to 0-1 using training set min and max
x_min = np.min(X_train, axis=0)
x_max = np.max(X_train, axis=0)
x_range = x_max - x_min
x_range[x_range == 0] = 1

X_train_n = (X_train - x_min) / x_range
X_test_n = (X_test - x_min) / x_range

# convert labels 1..7 to 0..6, then one-hot encode
y_train_oh = to_categorical(y_train - 1)
y_test_oh = to_categorical(y_test - 1)

def neural_network_classification(n_inputs, n_outputs):
    model = tf.keras.models.Sequential()
    model.add(Input((n_inputs,)))
    model.add(Dense(128, activation='relu'))
    model.add(Dense(64, activation='relu'))
    model.add(Dense(n_outputs, activation='softmax'))
    return model

model = neural_network_classification(X_train_n.shape[1], y_train_oh.shape[1])
model.summary()
model.compile(optimizer=tf.keras.optimizers.Adam(), loss='categorical_crossentropy', metrics=['accuracy'])

start_train = time.time()
history = model.fit(
    X_train_n, y_train_oh,
    epochs=25,
    batch_size=1000,
    verbose=2,
    validation_data=(X_test_n, y_test_oh)
)
end_train = time.time()

pred = model.predict(X_test_n)
pred_int = np.argmax(pred, axis=1) + 1

print('Elapsed time training = {:.4f} secs'.format(end_train-start_train))
print(f'{accuracy_score(pred_int, y_test) = :.4f}')
print('Confusion matrix')
print(confusion_matrix(y_test, pred_int))
```

```python
# Improved neural network attempt

# The improved neural network performed much better than the first neural-network attempt,
# but it still did not exceed the decision-tree baseline. This suggests that, for this
# dataset, the decision-tree baseline remained stronger than the dense neural network tested here.

import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.utils import to_categorical

# normalize inputs to 0-1 using training set min and max
x_min = np.min(X_train, axis=0)
x_max = np.max(X_train, axis=0)
x_range = x_max - x_min
x_range[x_range == 0] = 1

X_train_n = (X_train - x_min) / x_range
X_test_n = (X_test - x_min) / x_range

# convert labels 1..7 to one-hot
y_train_oh = to_categorical(y_train - 1)
y_test_oh = to_categorical(y_test - 1)

def neural_network_classification(n_inputs, n_outputs):
    model = tf.keras.models.Sequential()
    model.add(Input((n_inputs,)))
    model.add(Dense(256, activation='relu'))
    model.add(Dense(128, activation='relu'))
    model.add(Dense(64, activation='relu'))
    model.add(Dense(n_outputs, activation='softmax'))
    return model

model = neural_network_classification(X_train_n.shape[1], y_train_oh.shape[1])
model.summary()
model.compile(
    optimizer=tf.keras.optimizers.Adam(),
    loss='CategoricalCrossentropy',
    metrics=['accuracy']
)

start_train = time.time()
history = model.fit(
    X_train_n, y_train_oh,
    epochs=60,
    batch_size=500,
    verbose=2,
    validation_data=(X_test_n, y_test_oh)
)
end_train = time.time()

pred = model.predict(X_test_n)
pred_int = np.argmax(pred, axis=1) + 1

print('Elapsed time training = {:.4f} secs'.format(end_train-start_train))
print(f'{accuracy_score(pred_int, y_test) = :.4f}')
print('Confusion matrix')
print(confusion_matrix(y_test, pred_int))
```
