```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
```

```python
import zipfile
import os

# Define the path to the ZIP file and the destination directory
zip_file_path = 'datasets_Xy.zip'
extract_dir = 'datasets'

# Create the destination directory if it doesn't exist
os.makedirs(extract_dir, exist_ok=True)

# Open the ZIP file in read mode and extract all contents
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)
    print(f"Extracted all contents to {extract_dir}")
```

```python
marks = ['.r', '*b','+k']
for i in range(7):
  X = np.load(f'datasets/X{i}.npy')
  y = np.load(f'datasets/y{i}.npy')
  plt.figure()
  for c in range(np.max(y)+1):
    ind = y==c
    plt.plot(X[ind,0],X[ind,1],marks[c],label=f'y = {c}')
  plt.legend()
  plt.grid(True)
  plt.xlabel("a0 (X[:,0])")
  plt.ylabel("a1 (X[:,1])")
  plt.title(f'Dataset X{i},y{i}')

  model = DecisionTreeClassifier(criterion = 'entropy')
  model = model.fit(X, y)

  fig, ax = plt.subplots(figsize=(8,7))
  t = tree.plot_tree(model, feature_names=['a0','a1'],fontsize=9, ax=ax,class_names=[str(i) for i in range(np.max(y)+1)])
  plt.show()

  p = model.predict(X)
  print(f'{accuracy_score(p,y) = :.2f}')
  leaves = model.get_n_leaves()
  print(f'The tree has {leaves-1} internal nodes and {leaves} leaves')
  print(f"The tree's depth is {model.get_depth()}")
```

```python
n = 100
X = np.random.random(size = (n,2))
y = np.zeros(n,dtype=int)
y[n//2:] = 1

plt.figure()
for c in range(np.max(y)+1):
  ind = y==c
  plt.plot(X[ind,0],X[ind,1],marks[c],label=f'y = {c}')
plt.legend()
plt.grid(True)
plt.xlabel("a0 (X[:,0])")
plt.ylabel("a1 (X[:,1])")


model = DecisionTreeClassifier(criterion = 'entropy')
model = model.fit(X, y)


p = model.predict(X)
print(f'{accuracy_score(p,y) = :.2f}')
leaves = model.get_n_leaves()
print(f'The tree has {leaves-1} internal nodes and {leaves} leaves')
print(f"The tree's depth is {model.get_depth()}")
```
