```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import time
from sklearn.metrics import accuracy_score, confusion_matrix,mean_squared_error,mean_absolute_error
from sklearn.model_selection import train_test_split
from google.colab import files
from sklearn.neighbors import KNeighborsRegressor
```

```python
X_train = np.arange(-5,6)
y_train = X_train/2 + 1
plt.figure()
plt.plot(X_train, y_train,'.')
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
X_test = np.linspace(-5.5,5.5,200)
y_test = X_test/2 + 1

plt.figure()
plt.plot(X_test,y_test)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Evaluate the performance on k-nn on the dataset above.

```python
model = KNeighborsRegressor(n_neighbors=1)
model.fit(X_train.reshape(-1,1),y_train)
pred = model.predict(X_test.reshape(-1,1))
plt.figure()
plt.plot(X_train,y_train,'*',label='y_train')
plt.plot(X_test,pred,label='pred')
print('Mean squared error = {:5.3f}'.format(mean_squared_error(pred,y_test)))
print('Mean absolute error =  {:5.3f}'.format(mean_absolute_error(pred,y_test)))
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
