# **The dot product of two vectors**

The dot product of two vectors:

dotproduct.png

Observe that the dot product of vectors $x$ and $y$ is a scalar (single number) and is only defined for the case where  $x$ and $y$ have the same length.

We normally represent a vector as a 1D array.

```python
import numpy as np

x = np.array([2, 7, 1])
y = np.array([8, 2, 8])

print(f'{np.dot(x,y) = }')
```

```python
for i in range(2,5):
  x = np.random.randint(-1,4, size=i)
  y = np.random.randint(-1,4, size=i)
  print(f'np.dot({x},{y}) = {np.dot(x,y)}')
```

# **The product of two matrices**

Recall that a matrix is a rectangular or 2D array of numbers.



The product of matrices $X$ and $Y$ is a matrix containing the dot products of the rows of $X$ and the columns of $Y$, thus it is only defined when the number of columns of $X$ is the same as the number of rows of $Y$.  

Matrix_multiplication_diagram_2.svg

Thus if $C = A*B$, then $C_{i,j} $ is the dot product of row $i$ of $A$ and column $j$ of $B$.

In Python, the operation **np.matmul(A,B)** returns the product of matrices $A$ and $B$.

```python
A = np.random.randint(-1,4, size=(4,2))
B = np.random.randint(-1,4, size=(2,3))
print('A =')
print(A)
print('B =')
print(B)
print('A*B =')
print(np.matmul(A,B))
```

Due to braodcasting, matrix multiplication is equivalent to the dot product of two 2D arrays, provided their size are compatible.

```python
print(np.dot(A,B))
```

**Important:** Matrix multiplication is, in general, not commutative. Thus $A*B \neq B*A$.

**Important:** Matrix multiplication is a mathematical operation, it is not the same as array multiplication in numpy.



```python
A = np.array([[1,2],[3,4]])
B = np.array([[2,3],[5,6]])
print(A)
print(B)
print('A*B =')
print(A*B)
print('np.matmul(A,B) =')
print(np.matmul(A,B))
```

# **The identity matrix**

The identity matrix is a square matrix with ones on the main diagonal and zeros elsewhere.



```python
for n in range(1,5):
  print(f'np.identity({n}) =')
  print(np.identity(n))
```

Let $I_n$ denote an $n \times n$ identity matrix. If $A$ is an $r \times c$ matrix, then:

$ I_r* A = A$ and $ A * I_c = A$

In Python:


```python
r,c = 3,5
A = np.random.randint(-1,4, size=(r,c))
print('A =')
print(A)
I_c = np.identity(c,dtype=int)
print('I_c =')
print(I_c)
print('A*I_c =')
print(np.matmul(A,I_c))
I_r = np.identity(r,dtype=int)
print('I_r =')
print(I_r)
print('I_r*A =')
print(np.matmul(I_r,A))
```

# **The transpose of a matrix**

The transpose of a matrix $A$, denoted by $A^T$  is obtained by interchanging the rows and columns of $A$ such that $A_{i,j} = A^T_{j,i}$ for all valid indices $i$ and $j$.

```python
for i in range(3):
  r,c = np.random.randint(2,7,size=2)
  A = np.random.randint(-1,4, size=(r,c))
  print('A =')
  print(A)
  print('A.T =')
  print(A.T)
```

# **Inverse of a Matrix**

The inverse of a number $x$ is the number that, when multipled by $x$, gives $1$. Thus the inverse of $x$ is $1/x$, since $x *(1/x) = 1$.

Similarly, the inverse of a matrix $A$ is the matrix  that, when multiplied by $A$, gives an identity matrix. We denote the inverse of matrix $A$ as $A^{-1}$, thus $A * A^{-1} = I$.

Many libraries, including numpy, contain functions to compute the inverse of a matrix.

```python
np.set_printoptions(precision = 2)
np.random.seed(1)
A = np.random.randint(-1,4, size=(4,4))
print('A =')
print(A)
print('A**-1 =')
print(np.linalg.inv(A))
print('A * A**-1 =')
print(np.matmul(A,np.linalg.inv(A)))
```

```python
np.random.seed(1)
A = np.random.randint(-9,10, size=(4,4))
B = np.random.randint(-9,10, size=(4,4))
print('A =')
print(A)
print('B =')
print(B)
C = np.matmul(A,B)
print('C = A*B =')
print(C)
```

Can we find $A$ from $B$ and $C$?

Since

$$C = A*B$$

then

$$C * B^{-1} = A * B * B^{-1}$$

thus

$$C * B^{-1} = A * I$$

finally

$$C * B^{-1} = A$$

```python
print(np.linalg.inv(A))
print(np.linalg.inv(B))
print(np.linalg.inv(C))
```

```python
print('A =')
print(A)
print('C * B**-1 =')
print(np.matmul(C,np.linalg.inv(B)))
```

# **Pseudo-inverse of a Matrix**

The pseudo-inverse of a matrix $A$ is the matrix $A^*$ such that $A^**A$ is as close to the identity matrix as posible, that is, $ A^**A \approx I$.

The pseudoinverse of $A$ is given by:

$$A^* = (A^T*A)^{-1} * A^T$$



```python
np.random.seed(1)
A = np.random.randint(-9,10, size=(6,4))
print(A)

P = np.matmul(np.linalg.inv(np.matmul(A.T,A)),A.T)
print(P)

print('P * A =')
print(np.matmul(P,A))
```

Numpy provides an implementation of the pseudoinverse computation.

```python
np.random.seed(1)
A = np.random.randint(-9,10, size=(6,4))
print(A)

P = np.linalg.pinv(A)
print(P)

print('P * A =')
print(np.matmul(P,A))
```

# **Matrix Operations in Machine Learning**

A very common operation that appears in several machine learning models, including linear regression, logistic regression, and neural networks, consists of multiplying the input by a weight matrix $W$ and adding a bias vector $b$ to every row of the result.  

```python
n,a,h = 10,5,3

X = np.random.randint(-1,4, size=(n,a))
W = np.random.randint(-1,2, size=(a,h))
b = np.random.randint(-1,2, size=h)

print('X =')
print(X)
print('W =')
print(W)
print('b =')
print(b)

print('X*W =')
print(np.matmul(X,W))

print('X*W + b =')
print(np.matmul(X,W)+b)
```

Notice that the input $X$, of size $n\times a$, represents $n$ examples with $a$ attributes, and the output, of size $n\times h$, represents the same $n$ examples with $h$ attributes, thus we can view the columns in the output as new, higher-level attributes that describe the same original examples.



# **Matrix Operations in Machine Learning - an example**

```python
from sklearn.metrics import accuracy_score, confusion_matrix,mean_squared_error,mean_absolute_error
import tensorflow as tf

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
X_train = np.float32(X_train/255).reshape(X_train.shape[0],-1) # Convert every 28x28 image to a 784-element row
X_test = np.float32(X_test/255).reshape(X_test.shape[0],-1)    # Convert every 28x28 image to a 784-element row
y_train_oh = tf.keras.utils.to_categorical(y_train, num_classes = 10)
y_test_oh = tf.keras.utils.to_categorical(y_test, num_classes = 10)
print(f'Reshaped arrays: {X_train.shape=} {y_train_oh.shape=} {X_test.shape=} {y_test_oh.shape=}')
```

We want to find a matrix $W$ and a vector $b$ such that

$$X*W + b \approx y$$

where $(X,y)$ is our training set.

First we will let $b$ be the mean value (column-wise) of $y$, thus $b$ is a 1D array of size 10. Tha is: $$b = \mu(y)$$

Now we want to find $W$ such that
$$X*W \approx y - b$$

```python
print(f'{X_train.shape =}')
print(f'{y_train_oh.shape =}')
b = np.mean(y_train_oh,axis=0)
print(f'{b.shape =}')
print(f'{(y_train_oh-b).shape =}')
```

Multiplying both sides of the equation above by the pseudoinverse of $X$:

$$X^**X*W \approx X^** (y - b)$$

Since $X^**X \approx I$,

$$I*W \approx X^** (y - b)$$

finally

$$W \approx X^** (y - b)$$

Implemented in Python:

```python
b = np.mean(y_train_oh,axis=0)
W = np.matmul(np.linalg.pinv(X_train),(y_train_oh-b))
print(f'{b.shape =}')
print(f'{W.shape =}')
```

Now we can classify the test set using $W$ and $b$.

```python
pred = np.matmul(X_test,W) + b
print('Mean squared error = {:5.3f}'.format(mean_squared_error(pred,y_test_oh)))
print('Mean absolute error =  {:5.3f}'.format(mean_absolute_error(pred,y_test_oh)))
pred_int = np.argmax(pred,axis=1)
print('Accuracy = {:.4f}'.format(accuracy_score(pred_int,y_test)))
print('Confusion matrix')
print(confusion_matrix(y_test, pred_int))
```
